```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\P\Downloads\Customer Churn Prediction & Retention Prioritization\data\processed\cleaned_telco_churn.csv")
df.head()
```

```python
# Feature 1:Tenure_bucket
df["Tenure_bucket"] = pd.cut(df["tenure"],bins=[-1, 12, 24, 36, 48, 60, float("inf")],
labels=["0-12","13-24","25-36","37-48","49-60","60+"])
df['Tenure_bucket']
```

```python
# Feature 2: Total_service

service_columns = [
        "PhoneService",
        "OnlineSecurity",
        "OnlineBackup",
        "DeviceProtection",
        "TechSupport",
        "StreamingTV",
        "StreamingMovies"
    ]

df["total_services"] = 0

for column in service_columns:
    df["total_services"] += (
        df[column].eq("Yes").astype(int)
    )
```

```python
# Fetaure 3:Average_monthly_spend
df['Average_monthly_spend']=(df['TotalCharges']/df['tenure'].replace(0,1)).round(2)
```
